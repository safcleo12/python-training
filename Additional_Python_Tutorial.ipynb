👋 ⚒ Exercise 3: You want to count how many words in a sentence are longer than 4 letters. First restate the problem in your own words to make sure you understand the problem.

Then, manually count the words that have more than four letters. What is the expected result?

The draft of the code is the following:
1. Split the sentence into words
2. Create a counter variable that is initially 0
3. For each word, check if its length is bigger than four and if True increase counter by 1
4. Print the counter

```
sentence = "The quick brown fox jumps over the lazy dog"
counter = 0
words = sentence.split()
for word in words:
    if len(word) > 4:
        counter += 1



print(counter)

```

❓ Compare the  four versions in the following code cells. Is there any difference in the outcome?


```
fruits = ["apple", "banana", "cherry"]

# Version 1: manual counter
i = 0
for fruit in fruits:
    print(i, fruit)
    i += 1

# Version 2: range and indexing
for i in range(len(fruits)):
    print(i, fruits[i])

# Version 3: enumerate
for i, fruit in enumerate(fruits):
    print(i, fruit)

# Version 4: enumerate starting from 1 instead of 0
for i, fruit in enumerate(fruits, start=1):
    print(i, fruit)
```
